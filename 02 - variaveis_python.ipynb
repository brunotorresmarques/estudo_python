# Aula: Variáveis em Python 🐍

O que são variáveis, como criá-las, as regras de nomeação, conversão de tipos de dados e como trabalhar com entrada e saída de informações.

---
## 1. O que é uma Variável?

Uma **variável** é um espaço reservado na memória do computador usado para armazenar dados que podem ser modificados durante a execução do programa.

Pense em uma variável como uma **caixa etiquetada**: a etiqueta é o **nome** da variável e o conteúdo dentro da caixa é o **valor**.

```python
# Criando variáveis simples em Python
nome = "Ana"
idade = 25
altura = 1.68
estudante = True

# Exibindo os valores
print("Nome:", nome)
print("Idade:", idade)
print("Altura:", altura)
print("É estudante?", estudante)
```

---
## 2. Tipos Primitivos de Dados

O Python identifica automaticamente o tipo da variável com base no valor atribuído (tipagem dinâmica).

- `str` (String): Textos (sempre entre aspas)
- `int` (Integer): Números inteiros
- `float` (Float): Números decimais (ponto flutuante)
- `bool` (Boolean): Valores lógicos (`True` ou `False`)

Podemos verificar o tipo de uma variável com a função `type()`.

```python
# Verificando os tipos de dados com type()
print(type(nome))
print(type(idade))
print(type(altura))
print(type(estudante))
```

---
## 3. Regras de Nomeação de Variáveis

Para criar nomes válidos em Python:

1. **Podem conter:** letras, números e underline (`_`).
2. **Não podem começar com números.** (`1nome` ❌)
3. **São *Case-Sensitive*:** `Idade` e `idade` são variáveis diferentes.
4. **Não use palavras reservadas** da linguagem (como `if`, `for`, `print`, `def`).
5. **Boas Práticas:** Use o padrão `snake_case` (palavras em minúsculo separadas por `_`).

```python
# Nomes válidos (padrão recomendado)
nome_completo = "Maria Silva"
saldo_conta = 1500.50
ano_nascimento = 1998

# Atribuição múltipla
x, y, z = 10, 20, 30

print(nome_completo)
print(x, y, z)
```

---
## 4. Conversão de Tipos (Casting)

É possível converter o tipo de uma variável utilizando as funções de conversão: `str()`, `int()`, `float()` e `bool()`.

```python
# Exemplo de conversão
numero_texto = "100"
print("Tipo original:", type(numero_texto))

# Convertendo texto para inteiro
numero_inteiro = int(numero_texto)
print("Novo tipo:", type(numero_inteiro))

# Operação matemática após a conversão
print("Resultado:", numero_inteiro + 50)
```

---
## 5. Entrada de Dados (`input`)

A função `input()` permite receber dados informados pelo usuário. **Atenção:** o retorno do `input()` é sempre uma `str` (string).

```python
# Recebendo dados do usuário
usuario_nome = input("Digite o seu nome: ")
usuario_idade = int(input("Digite a sua idade: "))  # Convertendo diretamente para inteiro

# Exibindo mensagem personalizada usando f-strings
print(f"Olá, {usuario_nome}! Você tem {usuario_idade} anos.")
```

## 6. Exercício Prático ✏️

Escreva um programa que peça ao usuário:
1. O nome do produto.
2. O preço unitário (número decimal).
3. A quantidade comprada (número inteiro).

No final, calcule e exiba o **valor total** da compra.

```python
# Resolva o exercício abaixo:

produto = input("Digite o nome do produto: ")
preco = float(input("Digite o preço unitário: R$ "))
quantidade = int(input("Digite a quantidade: "))

total = preco * quantidade

print(f"\nResumo da compra:")
print(f"Produto: {produto}")
print(f"Total a pagar: R$ {total:.2f}")
```
